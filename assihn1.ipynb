```python

movies = ['Inception', 'Titanic', 'Avatar', 'Interstellar', 'The Dark Knight']
ratings = [9.0, 7.8, 8.2, 8.6, 9.1]


movie_ratings = dict(zip(movies, ratings))
print("Initial Dictionary:", movie_ratings)


print("\nMovies with rating above 8.5:")
for movie, rating in movie_ratings.items():
    if rating > 8.5:
        print(f"{movie}: {rating}")


movie_ratings['Titanic'] = 8.3

movie_ratings['Oppenheimer'] = 9.2
del movie_ratings['Avatar']
sorted_movies = dict(sorted(movie_ratings.items(), key=lambda x: x[1], reverse=True))

print("Sorted by ratings (descending):", sorted_movies)


highest_movie = max(movie_ratings, key=movie_ratings.get)
print(f"Highest Rated Movie: {highest_movie} ({movie_ratings[highest_movie]})")
```
